```python
location_missing = df.groupby('location').agg(
    total_records=('session_id', 'count'),
    missing_plugout=('plugout_time',
                     lambda x: x.isna().sum())
)

location_missing['missing_percentage'] = (
    location_missing['missing_plugout']
    / location_missing['total_records'] * 100
)

location_missing.sort_values(
    'missing_percentage',
    ascending=False
)
```

```python
df4_clean = df4.dropna(subset=['user_id', 'session_id']).copy()
```

```python
df4_clean.shape
```

```python
print(
    "Missing idle where connected equals charged:",
    (
        df4_clean['energy_idle_i'].isnull() &
        (df4_clean['energy_connected_i'] ==
         df4_clean['energy_charged_i'])
    ).sum()
)
```

```python
print(
    "Missing charged where connected equals idle:",
    (
        df4_clean['energy_charged_i'].isna() &
        (df4_clean['energy_connected_i'] ==
         df4_clean['energy_idle_i'])
    ).sum()
)
```
